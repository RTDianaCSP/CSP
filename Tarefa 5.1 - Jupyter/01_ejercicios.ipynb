<h2 style="color: #007BC4;">Ejercicio 1</h2>
<hr style="border: 0.5px solid #007BC4;">

Escribir un programa que lea un entero positivo, $n$, introducido por el usuario y después muestre en pantalla la suma de todos los enteros desde 1 hasta $n$. La suma de los $n$ primeros enteros positivos puede ser calculada mediante un bucle.


**Nota**: No es necesario comprobar que el número es positivo, se asume que se introducen los datos correctamente



#### Ejemplo de ejecución:

<pre>
Introduce un número positivo: 5
A suma dos números entre 1 e 5 é 15

--------------------------------------------------------------

Introduce un número positivo: 20
A suma dos números entre 1 e 20 é 210
</pre>

*Lo hice con un bucle ya que era más interesante. Ya se usa una fórmula en el siguiente ejercicio.*

```python
numero = int(input("Introduce un número positivo: "))
sum = 0

for i in range(1,numero+1): #para que sume tamén o propio número introducido
    sum = sum + i

print(f"A suma dos números entre 1 e {numero} é {sum}")

```

Escribir un programa que pida al usuario su peso (en kg) y estatura (en metros), calcule el índice de masa corporal y lo almacene en una variable, e imprima por pantalla la frase Tu índice de masa corporal es <imc> donde <imc> es el índice de masa corporal calculado redondeado con dos decimales.

#### Ejemplo de ejecución:

<pre>

<h2 style="color: #007BC4;">Ejercicio 3</h2>
<hr style="border: 0.5px solid #007BC4;">

Escribir un programa en Python que pida al usuario un tiempo en segundos (entero) y descomponga el valor en días, horas, minutos y
segundos transcurridos. Si algún valor es igual a cero se mostrará igualmente.
  
    
#### Ejemplo de ejecución:

<pre>
Introduce un tempo en segundos: 86465
Días:        1
Horas:       0
Minutos:     1
Segundos:    5
<pre>


```python
# Quitei prints innecesarios
tempo = int(input("Introduce un tempo en segundos: "))

d = 0
h = 0
m = 0
s = 0

if tempo > 0:
    while tempo >= 86400:
        tempo = tempo - 86400
        d = d + 1

    while tempo >= 3600:
        tempo = tempo - 3600
        h = h + 1

    while tempo >= 60:
        tempo = tempo - 60
        m = m + 1

    s = tempo #if tempo > 0: innecesario

    print(f"Días: {d}")
    print(f"Horas:  {h}")
    print(f"Minutos:  {m}")
    print(f"Segundos: {s}")
else:
    print("Introduce un tempo válido")


```

<h2 style="color: #007BC4;">Ejercicio 5</h2>
<hr style="border: 0.5px solid #007BC4;">

Implementar un programa en Python que calcule el número de vueltas que tendremos que darle a
un manguera para enrollarla en un soporte circular. Para ello:

- Se pedirá al usuario la longitud de la manguera y el radio del soporte en el que se enrollará, ambos en cm y números enteros.
- Se calculará el número de vueltas enteras que la manguera podría enrollarse sobre el carrete.
- Se calculará la longitud la longitud del trozo de manguera restante también en cm (número real).
  
    
#### Ejemplos de ejecución:

<pre>
Introduce a lonxitude da mangueira en cm: 100
Introduce o radio do soporte en cm: 5
O número de voltas foi 3 e sobran 5.75 cm
-------------------------------------------------------
Introduce a lonxitude da mangueira en cm: 150
Introduce o radio do soporte en cm: 10
O número de voltas foi 2 e sobran 24.34 cm
<pre>


```python
import math

l = int(input("Introduce a lonxitude da mangueira en cm: "))
r = int(input("Introduce o radio do soporte en cm: "))

e = 2*math.pi*r #non é necesario o float
sob = l #non é necesario o float
v = 0

while sob > e:
    v = v + 1
    sob = sob - e

print(f"O número de voltas foi {v} e sobran {sob:.2f} cm")
```
