### Exercicio 7: Lista de Diccionarios

Crea unha lista de diccionarios con información de estudantes:

```python
estudantes = [
    {"nome": "Ana", "idade": 20, "nota": 8.5},
    {"nome": "Carlos", "idade": 22, "nota": 7.8},
    {"nome": "María", "idade": 21, "nota": 9.2}
]
```

Realiza estas operacións:

1. Engade un novo estudante.
2. Atopa o estudiante con maior nota.
3. Calcula o promedio de idades.
4. Crea unha lista só cos nomes.
5. Ordena os estudantes por nota (descendente).

```python
estudantes.append({"nome": "Diana", "idade": 24, "nota":9})


mellor_nota = estudantes[0]
for e in estudantes:
  if(e["nota"]>mellor_nota["nota"]):
    mellor_nota = e
print("O estudante con maior nota é", mellor_nota["nome"], "con un", mellor_nota["nota"])

suma_idades=0
for e in estudantes:
  suma_idades+=e["idade"]
promedio_idades=suma_idades/len(estudantes)
print("Promedio de idades: ", promedio_idades)

lista_nomes = [e["nome"] for e in estudantes]
print("Lista cos nomes dos estudantes:", lista_nomes)

estudantes_ordenados = sorted(estudantes, key=lambda e: e["nota"], reverse=True)
print(estudantes_ordenados)



```

### Exercicio 8: Conxuntos

A partir destes conxuntos:

```python
conx_a = {1, 2, 3, 4, 5}
conx_b = {4, 5, 6, 7, 8}
```

Realiza estas operacións:

1. Unión de ambos conxuntos
2. Intersección
3. Diferencia (A - B)
4. Diferencia simétrica
5. Verifica si 3 está en A

```python
union = conx_a | conx_b
print("Unión:", union)

interseccion = conx_a & conx_b
print("Intersección:", interseccion)

diferenza = conx_a - conx_b
print("Diferenza:", diferenza)

diferenza_simetrica = conx_a ^ conx_b
print("Diferenza simétrica:", diferenza_simetrica)

print("Está 3 en A?", 3 in conx_a)
```

Crea un programa que:

### Exercicio 10: Combinación de Tipos Complexos

Temos un Sistema de biblioteca:

```python
biblioteca = {
    "libros": [
        {
            "titulo": "Cien años de soledad",
            "autor": "Gabriel García Márquez",
            "ano": 1967,
            "xeneros": {"Realismo mágico", "Novela"},
            "disponible": True
        },
        {
            "titulo": "1984",
            "autor": "George Orwell",
            "ano": 1949,
            "xeneros": {"Ciencia ficción", "Distopía"},
            "disponible": False
        }
    ],
    "usuarios": {
        "001": {"nome": "Ana", "libros_prestados": ["1984"]},
        "002": {"nome": "Carlos", "libros_prestados": []}
    }
}
```

Realiza estas operacións:

1. Engade un novo libro.
2. Marca un libro como non dispoñible.
3. Atopa todos os libros dun xénero.
4. Libros dispoñibles por autor.
5. Usuario con máis libros prestados.

```python
novo = {
            "titulo": "Frankstein",
            "autor": "Mary Shelley",
            "ano": 1818,
            "xeneros": {"Ciencia ficción", "Distopía"},
            "disponible": True
        }

novo2 = {
    "titulo": "El otoño del patriarca",
    "autor": "Gabriel García Márquez",
    "ano": 1975,
    "xeneros": {"Novela", "Realismo mágico"},
    "disponible": True
}


biblioteca["libros"].append(novo) #engadir novo libro
biblioteca["libros"].append(novo2)
print(biblioteca)

biblioteca["libros"][0]["disponible"] = False #marcar primeiro libro como non dispoñible
print(biblioteca)

print("Libros do xénero 'Ciencia ficción':") #atopar todos os libros dun xénero
for libro in biblioteca["libros"]:
  if "Ciencia ficción" in libro["xeneros"]:
    print("-",libro["titulo"], "de", libro["autor"])

#libros na biblioteca que estén dispoñibles por autor(que apareza na biblioteca)
autores_biblioteca = []
autores_sen_repetidos=[]
for libro in biblioteca["libros"]:

  if libro["autor"] not in autores_sen_repetidos:
    autores_sen_repetidos.append(libro["autor"])
  if libro["disponible"]==True:
    autores_biblioteca.append(libro["autor"])

print(autores_biblioteca)
print(autores_sen_repetidos)
for autor in autores_sen_repetidos:
  print(f"{autor} ten {autores_biblioteca.count(autor)} libros dispoñibles na biblioteca")

#Usuario con máis libros prestados
max_prestados = -1
usuario_max = ""

for usuario, info in biblioteca["usuarios"].items():
    cantidade = len(info["libros_prestados"])
    if cantidade > max_prestados:
        max_prestados = cantidade
        usuario_max = info["nome"]

if(max_prestados==-1):
  print("Ninguén ten ningún libro prestado")
else:
  print(f"O usuario con máis libros prestados foi {usuario_max} con {max_prestados} libros")







```
