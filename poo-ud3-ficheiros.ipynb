### Apertura Segura con try...except:

* Escribe unha función chamada `abrir_ficheiro_seguro(nome_ficheiro, modo)` que intente abrir un ficheiro no modo especificado.
* Debe usar un bloque `try...except` para capturar calquera excepción (como `FileNotFoundError`).
* Se a apertura é exitosa, a función debe devolver o obxecto do ficheiro. Se hai un erro, debe amosar a mensaxe de erro e devolver `None`.
* Fai probas invocando a función cun ficheiro que non exista e despois cun que si exista (por exemplo, en modo lectura `"r"`). Non esquezas pechar o ficheiro se se abriu correctamente.

````python
def abrir_ficheiro_seguro(nome_ficheiro, modo):
    try:
        ficheiro = open(nome_ficheiro, modo)
        print("Ficheiro aberto correctamente.")
        return ficheiro
    except Exception as e:
        print(f"Erro ao intentar abrir o ficheiro: {e}")
        return None

#proba cun que non exista
f1 = abrir_ficheiro_seguro("nonexiste.txt", "r")
print(f1)  # Debería ser None

#proba cun que existe
with open("existe.txt", "w") as f: #creoo antes
    f.write("Ola!")

# Probar apertura segura
f2 = abrir_ficheiro_seguro("existe.txt", "r")

if f2: #se se abriu pechase
    contido = f2.read()
    print("Contido do ficheiro:", contido)
    f2.close()  
````

### Diario de Notas Persoal

* Crea un programa que permita ao usuario escribir anotacións nun ficheiro chamado `diario.txt`.
* O programa debe:
    1. Preguntar ao usuario se quere (1) Ler todas as anotacións ou (2) Engadir unha nova anotación.
    2. Se escolle ler, debe abrir o ficheiro en modo lectura (`"r"`) e amosar todo o seu contido na pantalla. Controla o erro se o ficheiro aínda non existe.
    3. Se escolle engadir, debe pedirlle ao usuario unha liña de texto e **engadila** ao final do ficheiro (modo `"a"`), incluíndo a data e hora actual ao comezo da liña (usa o módulo `datetime`).


````python
from datetime import datetime

def ler_anotacions():
    try:
        with open("diario.txt", "r", encoding="utf-8") as f:
            contido = f.read()
            print("\n---Contido do diario---")
            print(contido)
    except FileNotFoundError:
        print("O diario aínda non existe.")

def engadir_anotacion():
    texto = input("Escribe unha anotación: ")
    tempo = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    
    with open("diario.txt", "a", encoding="utf-8") as f:
        f.write(f"[{tempo}] {texto}\n")
    
    print("Anotación gardada correctamente.")

def programa():
    continuar = True
    while(continuar):
        print("---Diario de Notas Persoal---")
        print("1.Ler todas as anotacións")
        print("2.Engadir unha nova anotación")
        print("3.Sair")
        
        opcion = input("Escoller opción (1, 2 ou 3): ")
    
        if opcion == "1":
            ler_anotacions()
        elif opcion == "2":
            engadir_anotacion()
        elif opcion == "3":
            continuar=False
        else:
            print("Opción non válida.")

programa()
````

###  Xestor de Contactos en JSON

* Crea un programa para xestionar unha axenda de contactos nun ficheiro de texto `contactos.JSON`.
* A estrutura será unha lista de dicionarios, onde cada dicionario ten as chaves `nome`, `telefono`, `email`.
* O programa debe ter un menú para:
    * **Engadir contacto**: Pide os datos e engádeos ao ficheiro.
    * **Buscar contacto**: Pide un nome e amosa o teléfono e email de todas as liñas que conteñan ese nome.
    * **Listar todos**: Amosa todos os contactos numerados.

````python
import json
import os

FICHEIRO_CONTACTOS = "contactos.json" 

contactos_proba =[
    {
        "nome": "Ana Somoza",
        "telefono": "600123456",
        "email": "ana@gmail.com"
    },
    {
        "nome": "Marcos Pérez",
        "telefono": "611987654",
        "email": "marcos@gmail.com"
    },
    {
        "nome": "María Suárez",
        "telefono": "622345678",
        "email": "maria@gmail.com"
    }
]

if not os.path.exists(FICHEIRO_CONTACTOS):
    with open(FICHEIRO_CONTACTOS, "w", encoding="utf-8") as f:
        json.dump(contactos_proba, f, ensure_ascii=False, indent=4)


def ler_contactos():
    if os.path.exists(FICHEIRO_CONTACTOS):
        with open(FICHEIRO_CONTACTOS, "r", encoding="utf-8") as f:
            try:
                return json.load(f)
            except json.JSONDecodeError:
                return []  #erro de json
    return []
    
def gardar_contactos(contactos):
    with open(FICHEIRO_CONTACTOS, "w", encoding="utf-8") as f:
        json.dump(contactos, f, ensure_ascii=False, indent=4)

def engadir_contacto():
    nome = input("Nome: ")
    telefono = input("Telefono: ")
    email = input("Email: ")
    
    contacto = {"nome": nome, "telefono": telefono, "email": email}
    contactos = ler_contactos()
    contactos.append(contacto)
    gardar_contactos(contactos)
    
    print("Contacto engadido correctamente!")

def buscar_contacto():
    nome_busca = input("Nome a buscar: ").lower()
    contactos = ler_contactos()
    
    encontrados = [c for c in contactos if nome_busca in c["nome"].lower()]
    
    if encontrados:
        print("\n---Contactos encontrados---")
        for c in encontrados:
            print(f"Nome: {c['nome']}, Teléfono: {c['telefono']}, Email: {c['email']}")
    else:
        print("Non se atopou ningún contacto.")

def listar_todos():
    contactos = ler_contactos()
    if contactos:
        print("\n---Todos os contactos---")
        for i, c in enumerate(contactos, start=1):
            print(f"{i}. Nome: {c['nome']}, Teléfono: {c['telefono']}, Email: {c['email']}")
    else:
        print("A axenda está baleira.")


def xestor():
    continuar = True
    
    while continuar:
        print("\n---XESTOR DE CONTACTOS---")
        print("1. Engadir contacto")
        print("2. Buscar contacto")
        print("3. Listar todos")
        print("4. Saír")
        
        opcion = input("Escolle unha opción (1-4): ")
        
        if opcion == "1":
            engadir_contacto()
        elif opcion == "2":
            buscar_contacto()
        elif opcion == "3":
            listar_todos()
        elif opcion == "4":
            continuar=False
        else:
            print("Opción non válida. Proba de novo.")

xestor()
````

### Cálculo con Datos Numéricos usando NumPy

    1.5 2.3 4.8

````python
import numpy as np

with open("matriz.txt", "w", encoding="utf-8") as f:
    f.write("1.5 2.3 4.8\n")
    f.write("5.1 3.2 7.4\n")
    f.write("9.6 8.7 6.5\n")

try:
    datos = np.loadtxt("matriz.txt")
except Exception as e:
    print(f"Erro ao ler o ficheiro: {e}")
    exit()

media= np.mean(datos)
suma_filas = np.sum(datos, axis=1)
max_columnas = np.max(datos, axis=0)

print(f"Media de todos os números: {media:.2f}")
print(f"Suma de cada fila: {suma_filas}")
print(f"Valor máximo de cada columna: {max_columnas}")

datos_doble = datos * 2
np.savetxt("matriz_doble.txt", datos_doble, fmt="%.2f")

try:
    with open("matriz_doble.txt", "r", encoding="utf-8") as f:
        print("---matriz_doble.txt---")
        for liña in f:
            print(liña.strip()) 
except FileNotFoundError:
    print("Erro: Non existe")
````
